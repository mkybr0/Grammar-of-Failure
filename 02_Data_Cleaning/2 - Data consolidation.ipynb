# Data Consolidation

## 1. Import Packages & Define Custom Functions

```python
import pandas as pd
import re
```

```python
# Using custom function from Michael Albert's class

def summarize_dataframe(df):
    missing_values = pd.concat([pd.DataFrame(df.columns, columns=['Variable Name']), 
                      pd.DataFrame(df.dtypes.values.reshape([-1,1]), columns=['Data Type']),
                      pd.DataFrame(df.isnull().sum().values, columns=['Missing Values']), 
                      pd.DataFrame([df[name].nunique() for name in df.columns], columns=['Unique Values'])], 
                     axis=1).set_index('Variable Name')
    return pd.concat([missing_values, df.describe(include='all').transpose()], axis=1).fillna("")
```

## 2. Import Data

### 2a. Import Original Raw Data

```python
data_full = pd.read_csv('Raw Input Files/asrs_full.csv')
```

### 2b. Import Cleaned Data & Dummies

```python
data_anomaly_code = pd.read_csv('Output Files/Data 1 - anomaly clean.csv')
data_location_id = pd.read_csv('Output Files/Data 2 - location_id clean.csv')
data_aircraft_type = pd.read_csv('Output Files/Data 3 - aircraft_type clean.csv')
data_reporter_function = pd.read_csv('Output Files/Data 4 - reporter_function clean.csv')
data_flight_phase = pd.read_csv('Output Files/Data 5 - flight_phase clean.csv')
data_severity_scores = pd.read_csv('Output Files/Data 6 - severity_scores.csv')
data_addtl_ASRS = pd.read_csv('Output Files/Data 7 - addtl_ASRS clean.csv')
```

## 3. Concatenate Data

```python
data_concat = data_full.copy()
```

```python
data_concat.info()
```

```python
#Concat severity scores
data_concat = pd.concat([data_concat, data_severity_scores.drop('index', axis=1)], axis=1)
```

```python
# Concat anomaly_code
data_concat = pd.concat([data_concat.drop('anomaly_code', axis=1), data_anomaly_code.drop('acn', axis=1)], axis=1)
```

```python
# Concat location_id
data_concat = pd.concat([data_concat.drop('location_id', axis=1), data_location_id.drop('acn', axis=1)], axis=1)
```

```python
# Concat aircraft_type
data_concat = pd.concat([data_concat.drop('aircraft_type', axis=1), data_aircraft_type.drop('acn', axis=1)], axis=1)
```

```python
# Concat reporter_function
data_concat = pd.concat([data_concat.drop('reporter_function', axis=1), data_reporter_function.drop('acn', axis=1)], axis=1)
```

```python
# Concat flight_phase
data_concat = pd.concat([data_concat.drop('flight_phase', axis=1), data_flight_phase.drop('acn', axis=1)], axis=1)
```

```python
# Concat addtl_ASRS
data_concat = pd.concat([data_concat, data_addtl_ASRS.drop(['acn'], axis=1)], axis=1)
```
