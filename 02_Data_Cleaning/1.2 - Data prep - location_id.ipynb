```python
# Load your data
df = pd.read_csv('Intermediate Input Files/locationsplit.csv')

# Load OurAirports data
airports = pd.read_csv('Raw Input Files/airports.csv')

# Add municipality to the lookup
lookup = airports[['ident', 'iata_code', 'iso_country', 'iso_region', 'municipality']].copy()

# Build mapping dicts
icao_map = lookup.set_index('ident')[['iso_country', 'iso_region', 'municipality']].to_dict('index')
iata_map = lookup.dropna(subset=['iata_code']).set_index('iata_code')[['iso_country', 'iso_region', 'municipality']].to_dict('index')
```

```python
def get_country_name(iso_code):
    try:
        return pycountry.countries.get(alpha_2=iso_code).name
    except:
        return iso_code  # fall back to ISO code if not found

def get_location(code):
    if pd.isna(code):
        return None, None, None
    match = icao_map.get(code) or iata_map.get(code)
    if not match:
        return None, None, None
    country = match['iso_country']
    region = match['iso_region']
    city = match['municipality']
    if country == 'US':
        return 'United States', region.split('-')[-1], city
    return get_country_name(country), None, city

df[['country', 'state', 'city']] = df['airport'].apply(
    lambda x: pd.Series(get_location(x))
)
```

### 3.2 Consolidate Airport and Organization_Unit Values

```python
df = pd.read_csv('Intermediate Input Files/LocationID.csv')
```

```python
airport_value_counts = pd.DataFrame(df['airport'].value_counts().reset_index())

with pd.option_context('display.max_rows', None):
    display(airport_value_counts)
```

```python
print(len(airport_value_counts))
```

```python
# Filtering to keep all those airports that appear at least 30 times in order to follow sampling size rule of thumb
discard_airports = airport_value_counts[90:]['airport']
discard_airports
```

```python
# Filtering
df['airport'] = df['airport'].where(~df['airport'].isin(discard_airports), 'other')

# Viewing
with pd.option_context('display.max_rows', None):
    display(df['airport'].value_counts())
```

```python
with pd.option_context('display.max_rows', None):
    display(df['organizational_unit'].value_counts())
```

```python
# Consolidating VOR and VORTAC

df['organizational_unit'] = df['organizational_unit'].replace(
    {'VORTAC': 'VOR/VORTAC', 'VOR': 'VOR/VORTAC'}
)

# Treating other Rare Values

df['organizational_unit'] = df['organizational_unit'].replace(
    {'NDB': 'other', 'FSS': 'other', 'TACAN':'other', 'Airport; AKR.Airport': 'other'}
)

with pd.option_context('display.max_rows', None):
    display(df['organizational_unit'].value_counts())
```

### 3.3. Encode Data

```python

# Dummy code airport
airport_dummies = pd.get_dummies(df['airport'], prefix = 'airport').astype(int)
df = pd.concat([df, airport_dummies], axis = 1)

# Dummy code organizational_unit
org_dummies = pd.get_dummies(df['organizational_unit'], prefix = 'org').astype(int)
df = pd.concat([df, org_dummies], axis = 1)

# One-hot encode state and airport_type
state_dummies = pd.get_dummies(df['state'], prefix='state').astype(int)
airport_type_dummies = pd.get_dummies(df['airport_type'], prefix='airport_type').astype(int)

# Insert state dummies after state column
state_idx = df.columns.get_loc('state') + 1
for col in reversed(state_dummies.columns):
    df.insert(state_idx, col, state_dummies[col])

# Insert airport_type dummies after airport_type column
airport_type_idx = df.columns.get_loc('airport_type') + 1
for col in reversed(airport_type_dummies.columns):
    df.insert(airport_type_idx, col, airport_type_dummies[col])
```
